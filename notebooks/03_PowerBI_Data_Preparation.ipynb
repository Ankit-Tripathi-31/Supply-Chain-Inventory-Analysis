```python

import pandas as pd
from pathlib import Path

project_path = Path(
    r"C:\Users\ANKIT TRIPATHI\Supply-Chain-Inventory-Analysis"
)

data_path = project_path / "data" / "Supply_Chain_Cleaned.csv"

df_powerbi = pd.read_csv(data_path)

df_powerbi["Order_Date"] = pd.to_datetime(
    df_powerbi["Order_Date"]
)

df_powerbi["Delivery_Date"] = pd.to_datetime(
    df_powerbi["Delivery_Date"]
)

print("Dataset loaded successfully!")
print("Dataset shape:", df_powerbi.shape)
print("\nColumn names:")
print(df_powerbi.columns.tolist())

display(df_powerbi.head())
```

```python

# Classify inventory status

df_powerbi["Inventory_Status"] = "Normal"

df_powerbi.loc[
    df_powerbi["Stock_Quantity"] <= df_powerbi["Reorder_Level"],
    "Inventory_Status"
] = "At or Below Reorder Level"

print("Inventory status column added successfully!")

display(
    df_powerbi[
        [
            "Product_Name",
            "Stock_Quantity",
            "Reorder_Level",
            "Inventory_Status"
        ]
    ].head(10)
)

print("\nInventory status counts:")
display(df_powerbi["Inventory_Status"].value_counts())
```

```python

print(df_powerbi.columns.tolist())
```

```python

# Recreate calculated columns safely

df_powerbi["Total_Gross_Profit"] = (
    df_powerbi["Profit_Per_Unit"] *
    df_powerbi["Units_Sold"]
)

df_powerbi["Inventory_Status"] = "Normal"

df_powerbi.loc[
    df_powerbi["Stock_Quantity"] <= df_powerbi["Reorder_Level"],
    "Inventory_Status"
] = "At or Below Reorder Level"

print("Calculated columns created successfully!")

display(
    df_powerbi[
        [
            "Product_Name",
            "Units_Sold",
            "Profit_Per_Unit",
            "Total_Gross_Profit",
            "Inventory_Status"
        ]
    ].head()
)
```

```python

print("Dataset shape:", df_powerbi.shape)

print("Missing values:", df_powerbi.isnull().sum().sum())
print("Duplicate rows:", df_powerbi.duplicated().sum())

print(
    "Gross profit calculation valid:",
    (
        df_powerbi["Total_Gross_Profit"]
        == df_powerbi["Profit_Per_Unit"] * df_powerbi["Units_Sold"]
    ).all()
)

print("\nInventory status counts:")
display(df_powerbi["Inventory_Status"].value_counts())
```

```python

# Export final dataset for Power BI

output_path = (
    project_path / "data" / "Supply_Chain_PowerBI.csv"
)

df_powerbi.to_csv(output_path, index=False)

print("Power BI dataset exported successfully!")
print("Saved location:", output_path)
print("Rows:", df_powerbi.shape[0])
print("Columns:", df_powerbi.shape[1])
print("File exists:", output_path.exists())
```
